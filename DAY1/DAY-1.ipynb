```python
### llm_obj.invoke('UserQuery') <== (A)
#   ------- ====================
#
#    Vs
#   prompt_object | llm_object  <== Chain - LangChain Expression (LCE)
#   -----(1)-----   ---(2)----

# prompt -> llm //chain

#  chain.invoke({'placeholder':'UserInput'})  <== (B)
# -------

##   (A) Vs (B)
#-----------------------------------------------------------
```

```python
# created llm object
#-----------------------
llm_obj = ChatCohere(model = "command-a-03-2025",cohere_api_key=COHERE_API_KEY)
```

```python
# promptObject
#---------------------
pobj = PromptTemplate(input_variables=['topic','level','n'],template="Explain about {topic} for a {level} student in {n} points")
####
```

```python
chain = pobj|llm_obj
```

```python
chain
```

```python
chain.invoke({'topic':'Python','level':'beginner','n':3})
```

```python
response = chain.invoke({'topic':'Python','level':'beginner','n':3})
print(response.content)
```
