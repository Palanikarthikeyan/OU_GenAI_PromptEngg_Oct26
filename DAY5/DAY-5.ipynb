```python
# Step - 1 : Configure LLM Object
Settings.llm = Cohere(model="command-a-03-2025",api_key=COHERE_API_KEY)

# Step - 2: Create Cohere Embedding
Settings.embed_model = CohereEmbedding(api_key=COHERE_API_KEY,model_name="embed-english-v3.0",input_type="search_document")

# Step -3 : Load PDF
documents = SimpleDirectoryReader("./data").load_data()
print("PDF Loaded Successfully")
# Step -4: Create Index
index = VectorStoreIndex.from_documents(documents)
print("Index Created Successfully")

# Step -5: Create Query Engine - similarity
query_engine = index.as_query_engine(similarity_top_k = 3)

# Step-6 : Define Custom Prompt
# --------------------------------
qa_prompt = PromptTemplate(
    """
    You are a helpful AI assistant.
    Use the following context to answer the
    question.
    Context:
    {context_str}

    Question:
    {query_str}
    Instructions:
    1. Explain in simple english
    2. Give three important points
    3. If the answer is not the context, say
    "Information not available in the documents".
    Answer:
    """
)
query_engine.update_prompts({"text_qa_template":qa_prompt})
```

```python
# Step-7: Ask question
#-------------------------
response = query_engine.query("what is self-attention")
print(response)
```

```python
def f1(arg): # decorator
  def f2(): # wrapper
    arg()
  return f2

def fx():
  print("City-1 news page")

fx = f1(fx)
fx()
```

```python
@f1
def fy():
  '''this is about City-2 news page'''
  print("City-2 news page")
  return 'City-2 news page'

fy()
```
