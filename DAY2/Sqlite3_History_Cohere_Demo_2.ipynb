```python
#############
# Complete Chathistory code
# -------------------------

# 1. Create the LLM model object

llm = ChatCohere(
    model="command-a-03-2025",
    cohere_api_key=COHERE_API_KEY
)


# 2. Create a prompt that accept conversation history
# -------------------------------------------------------

prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful GenAI training assistant"),
    MessagesPlaceholder(variable_name="history"),
    ("human","{question}")
])

# 3. Create prompt and llm chain  - LEC
# ------------------------------------------
chain = prompt | llm

# 4. Store histories separately for each session
# --------------------------------------------------
stores = {}
def get_session_history(session_id: str) -> BaseChatMessageHistory:
  if session_id not in stores:
    stores[session_id] = ChatMessageHistory()
  return stores[session_id]


# 5. Automatically manage conversation history
# -----------------------------------------------
chatbot = RunnableWithMessageHistory(chain,get_session_history,input_messages_key="question",history_messages_key="history")


# 6. First Question
# --------------------------
response1 = chatbot.invoke(
    {'question':'My name is Raj. I am learning GenAI'},
    config = {'configurable':{'session_id':'student_101'}}
)
```

```python
'''
User asks a question
     |
     |
----------------------------
RunnableWithMessageHistory
Uses the session ID to identify the conversation
-----------------------------------
     |
SQLite database
Fetches earlier human and AI messages
    |
Prompt + History + Current Question
Sent to Cohere through your existing chain
    |
    |
Cohere LLM response
The response is saved in SQLite for the next turn
'''
```

```python

import os
import sqlite3

from langchain_cohere import ChatCohere

from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.messages import (
    messages_from_dict,
    messages_to_dict,
)
from langchain_core.prompts import (
    ChatPromptTemplate,
    MessagesPlaceholder,
)
from langchain_core.runnables.history import (
    RunnableWithMessageHistory,
)


# --------------------------------------------------
# 1. Configure the database
# --------------------------------------------------

DB_PATH = "chat_history.db"


# --------------------------------------------------
# 2. Implement SQLite-based chat history
# --------------------------------------------------

class SQLiteChatMessageHistory(BaseChatMessageHistory):

    def __init__(self, session_id: str, db_path: str = DB_PATH):
        self.session_id = session_id
        self.db_path = db_path

        # Create the table if it does not exist
        with sqlite3.connect(self.db_path) as conn:
            conn.execute("""
                CREATE TABLE IF NOT EXISTS chat_messages (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    session_id TEXT NOT NULL,
                    message_json TEXT NOT NULL
                )
            """)

    @property
    def messages(self):
        """Fetch conversation history for this session."""

        with sqlite3.connect(self.db_path) as conn:
            rows = conn.execute("""
                SELECT message_json
                FROM chat_messages
                WHERE session_id = ?
                ORDER BY id ASC
            """, (self.session_id,)).fetchall()

        # Convert JSON strings back into LangChain messages
        serialized_messages = [
            __import__("json").loads(row[0])
            for row in rows
        ]

        return messages_from_dict(serialized_messages)

    def add_messages(self, messages):
        """Save human and AI messages to SQLite."""

        serialized_messages = messages_to_dict(messages)

        with sqlite3.connect(self.db_path) as conn:
            conn.executemany("""
                INSERT INTO chat_messages (session_id, message_json)
                VALUES (?, ?)
            """, [
                (
                    self.session_id,
                    __import__("json").dumps(message)
                )
                for message in serialized_messages
            ])

    def clear(self):
        """Delete history for this session only."""

        with sqlite3.connect(self.db_path) as conn:
            conn.execute("""
                DELETE FROM chat_messages
                WHERE session_id = ?
            """, (self.session_id,))


# --------------------------------------------------
# 3. Create the Cohere LLM
# --------------------------------------------------

#COHERE_API_KEY = os.environ.get("COHERE_API_KEY")
COHERE_API_KEY = userdata.get('COHERE_API_KEY')

llm = ChatCohere(
    model="command-a-03-2025",
    cohere_api_key=COHERE_API_KEY
)


# --------------------------------------------------
# 4. Create a prompt that accepts conversation history
# --------------------------------------------------

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful GenAI training assistant."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{question}")
])


# --------------------------------------------------
# 5. Create the LLM chain
# --------------------------------------------------

chain = prompt | llm


# --------------------------------------------------
# 6. Return SQLite history for each session
# --------------------------------------------------

def get_session_history(session_id: str) -> BaseChatMessageHistory:
    return SQLiteChatMessageHistory(
        session_id=session_id,
        db_path=DB_PATH
    )


# --------------------------------------------------
# 7. Automatically manage conversation history
# --------------------------------------------------

chatbot = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="question",
    history_messages_key="history"
)


# --------------------------------------------------
# 8. First question
# --------------------------------------------------

response1 = chatbot.invoke(
    {"question": "My name is Raj. I am learning GenAI."},
    config={
        "configurable": {
            "session_id": "student_101"
        }
    }
)

print("Response 1:", response1.content)


# --------------------------------------------------
# 9. Second question - same session
# --------------------------------------------------

response2 = chatbot.invoke(
    {"question": "What is my name, and what am I learning?"},
    config={
        "configurable": {
            "session_id": "student_101"
        }
    }
)

print("Response 2:", response2.content)


# --------------------------------------------------
# 10. Display messages stored in SQLite
# --------------------------------------------------

history = get_session_history("student_101")

print("\n--- Conversation History from SQLite ---")

for message in history.messages:
    print(f"{message.type}: {message.content}")
```
