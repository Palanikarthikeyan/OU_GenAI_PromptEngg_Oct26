```python
# Activity-1
# ----------------
'''
Create a resuable prompt template that acts as a Python trainer and explains programming
concepts based on user input
Task:
Create a chatbot that explains any python concept using:
  1. A simple definition
  2. A pratical python example
  3. An explanation of code
  4. One practice question
'''
from langchain_cohere import ChatCohere
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


# Create an LLM Object
# ---------------------
llm = ChatCohere(
    model="command-a-03-2025",
    cohere_api_key=COHERE_API_KEY
)


# Prompt Object
# -------------------

prompt = ChatPromptTemplate.from_messages([
    ("system","""
    You are an experienced Python trainer.
    Explain concepts in simple language.
    Include a definition,python example,code explanation
    and practice question
    """),
    ("human","Explain the Python concept:{topic}")
])

output_obj = StrOutputParser()

chain = prompt | llm |output_obj

response = chain.invoke({'topic':'Python inheritance'})
print(response)
```

```python
# Activit-2
# ----------
# Customer support AI Assistant
'''
Build Customer support assistant that drafts professional
replies based on a customer's issue and the desired tone.
Task:
Create an assistant that accepts:
1. Customer issues
2. Product name
3. Response tone
The assistant must generate an appropriate customer-support
reply
'''
from langchain_cohere import ChatCohere
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


# Create an LLM Object
# ---------------------
llm = ChatCohere(
    model="command-a-03-2025",
    cohere_api_key=COHERE_API_KEY
)


# Prompt Object
# -------------------

prompt = ChatPromptTemplate.from_messages([
    ("system","""
    You are a professional customer support assistant.
    Write clear,polite and helpful responses.
    If information is missing ask for clarification
    """),
    ("human","""
    Product:{product}
    Customer issue:{issue}
    Tone:{tone}

    Draft a suitable customer support reply.
    """)
])

output_obj = StrOutputParser()

chain = prompt | llm |output_obj

```

```python
#############
# Complete Chathistory code
# -------------------------

# 1. Create the LLM model object

llm = ChatCohere(
    model="command-a-03-2025",
    cohere_api_key=COHERE_API_KEY
)


# 2. Create a prompt that accept conversation history
# -------------------------------------------------------

prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful GenAI training assistant"),
    MessagesPlaceholder(variable_name="history"),
    ("human","{question}")
])

# 3. Create prompt and llm chain  - LEC
# ------------------------------------------
chain = prompt | llm

# 4. Store histories separately for each session
# --------------------------------------------------
stores = {}
def get_session_history(session_id: str) -> BaseChatMessageHistory:
  if session_id not in stores:
    stores[session_id] = ChatMessageHistory()
  return stores[session_id]


# 5. Automatically manage conversation history
# -----------------------------------------------
chatbot = RunnableWithMessageHistory(chain,get_session_history,input_messages_key="question",history_messages_key="history")


# 6. First Question
# --------------------------
response1 = chatbot.invoke(
    {'question':'My name is Raj. I am learning GenAI'},
    config = {'configurable':{'session_id':'student_101'}}
)
```
